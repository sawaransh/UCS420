1. Take your roll number. Extract its individual digits, and multiply each digit by 10 to form your list L (10 numbers in list now).

    * **i. Print L**
    * **ii. Add two numbers of your choice to L** - one using `append()` and one using `insert()` at a specific position.
    * **iii. Print L after each and explain in a comment what changed.**
    * **iv. Remove two elements from L** — one using `remove()` and one using `pop()`. Print L after each.
    * **v. Sort L in ascending order using `sort()`, then descending order using `sort(reverse=True)`.** Print both.
    * **vi. Use slicing to print the first three elements and the last three elements of L** in a single line each.
    * **vii. Use list comprehension to create a new list containing only the elements of L that are greater than the average of L.**

````python
num = int(input("Enter your Roll Number: "))
digits = [int(digit)*10 for digit in str(num)]
print(digits)

````

````python
# The append command will append the digit 7 to the end of the list
digits.append(7)
print(digits)

# The insert command will insert a digit to the list at a specific index by adding a new element at that index
digits.insert(0,18)
````

````python
# The pop command will delete the element at the specified index by removing it completely and shortening the list
digits.pop(0)
print(digits)

# The remove command will remove a specific list member by matching its exact value
digits.remove(7)
print(digits)
````

````python
# The sort comand will sort the list in ascending order
digits.sort()
print(digits)
````

````python
# Using the slicing technique to print the first three and last three members of the list
print(digits[:3])
print(digits[-3:])
````

````python
# Using list comprehension to create a list of the elements that are greater than average

average = sum(digits)/len(digits)
new_list = [ x for x in digits if x > average]
print(new_list)
````

3. Set the random seed to your own roll number using `random.seed(your_roll_number)` before generating your numbers — this keeps your list random-looking but reproducible and unique to you (a different seed produces a different list, so this cannot be copied from a classmate).

    * **i. Generate a list of 100 random numbers between 100 and 900** (inclusive).
    * **ii. Count and print all odd numbers in the list.**
    * **iii. Count and print all even numbers in the list.**
    * **iv. Count and print all prime numbers in the list**, and additionally build the actual list of prime numbers found using list comprehension.
    * **v. Print the number that occurs most frequently in your list**, and how many times it occurs.

````python
import random as rd

rd.seed(1024160030)

# i. Generate 100 random numbers between 100 and 900
numbers = [rd.randint(100,900) for x in range(100)]
print(numbers)

# ii. Count and print all odd numbers

odd_numbers = [ x for x in numbers if x%2 != 0 ]
print("Odd Numbers: ", odd_numbers)
print("Number of odd numbers:", len(odd_numbers))

# iii. Count and print all even numbers

even_numbers = [ x for x in numbers if x%2 == 0]
print("Even Numbers: ", even_numbers)
print("Number of even numbers: ", len(even_numbers))

#function to check for prime numbers

def is_prime(n):
    if n < 2:
      return False

    for i in range(2, int(n**0.5) + 1):
      if n % i == 0:
        return False

    return True
# iv. Find prime numbers using list comprehension
prime_numbers = [x for x in numbers if is_prime(x)]

print("\nPrime numbers:")
print(prime_numbers)
print("Number of prime numbers:", len(prime_numbers))

# v. Find the most frequently occurring number
most_frequent = max(set(numbers), key=numbers.count)
frequency = numbers.count(most_frequent)

print("\nMost frequently occurring number:", most_frequent)
print("It occurs", frequency, "time(s)")


````

4. Using the same 8 digits of your roll number from Q1 (before multiplying by 10), create set A = `{digit × 7 for each digit}` and set B = `{digit × 9 for each digit}`. Print both sets.

    * **i. Find the union of A and B** — all unique values across both sets.
    * **ii. Find the intersection of A and B** — values common to both.
    * **iii. Find A − B and B − A separately (not combined) using `difference()`,** and explain in one line how this differs from `symmetric_difference()`.
    * **iv. Find the symmetric difference of A and B.**
    * **v. Check whether A is a subset of B, and whether B is a superset of A,** using `issubset()` and `issuperset()`.
    * **vi. Ask the user to input a value X.** Remove X from set A if present using `discard()` (not `remove()`) — explain in one line why `discard()` is safer than `remove()` when you're not sure the value exists.

````python
# 8 digits of roll number
digits = [1, 2, 3, 4, 5, 6, 7, 8]

# Create sets A and B
A = {digit * 7 for digit in digits}
B = {digit * 9 for digit in digits}

# Print both sets
print("Set A:", A)
print("Set B:", B)

# vi. Union of A and B
union = A.union(B)
print("\nUnion of A and B:", union)

# vii. Intersection of A and B
intersection = A.intersection(B)
print("Intersection of A and B:", intersection)

# viii. A - B and B - A
A_minus_B = A.difference(B)
B_minus_A = B.difference(A)

print("A - B:", A_minus_B)
print("B - A:", B_minus_A)

# difference() gives values present only in the first set,
# while symmetric_difference() gives values that are in either set but not both.

# ix. Symmetric difference
symmetric_diff = A.symmetric_difference(B)
print("Symmetric difference:", symmetric_diff)

# x. Check subset and superset
print("Is A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))

# xi. Ask user for a value X
X = int(input("\nEnter a value X: "))

# discard() removes X if it exists, without giving an error if it doesn't.
A.discard(X)

print("Set A after discarding X:", A)

# discard() is safer than remove() because it does not raise an error
# if the value is not present in the set.
````

````python
# Original dictionary

my_dict = {
    "name": "Sawaransh",
    "roll_no": "YOUR_ROLL_NUMBER",
    "branch": "COSE",
    "age": 20,
    "city": "Your Home City"
}

print("Original dictionary:")
print(my_dict)


# i. Rename "city" to "location" using pop()

my_dict["location"] = my_dict.pop("city")

print("\nAfter renaming city to location:")
print(my_dict)


# ii. Add a new key "cgpa"

my_dict["cgpa"] = 8.5

print("\nAfter adding CGPA:")
print(my_dict)


# iii. Increase age by 1

my_dict["age"] = my_dict["age"] + 1

print("\nAfter increasing age by 1:")
print(my_dict)


# iv. Delete "branch" using pop() and del
# Create two separate copies

dict_pop = my_dict.copy()
dict_del = my_dict.copy()

# Using pop()
removed_branch = dict_pop.pop("branch")

print("\nDictionary after deleting branch using pop():")
print(dict_pop)
print("Value returned by pop():", removed_branch)

# Using del
del dict_del["branch"]

print("\nDictionary after deleting branch using del:")
print(dict_del)

# pop() removes and returns the value,
# while del removes the key-value pair but does not return it.


# v. Iterate using .items()

print("\nKey-value pairs:")

for key, value in my_dict.items():
    print(key, "→", value)


# vi. Check whether "email" exists before accessing it

if "email" in my_dict:
    print("\nEmail:", my_dict["email"])
else:
    print("\nEmail not found in dictionary.")


# vii. Create friend's dictionary

friend_dict = {
    "name": "Rahul",
    "roll_no": "FRIEND123",
    "branch": "COSE",
    "age": 21,
    "city": "Delhi"
}

# Merge the two dictionaries

merged_dict = {**my_dict, **friend_dict}

print("\nMerged dictionary:")
print(merged_dict)

# When both dictionaries have the same key,
# the value from the second dictionary wins.


# viii. Dictionary comprehension to keep only string values

string_values = {
    key: value
    for key, value in my_dict.items()
    if isinstance(value, str)
}

print("\nDictionary containing only string values:")
print(string_values)
````
