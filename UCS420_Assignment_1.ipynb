**Assingment 5.1:** WAP to find max amoung three numbers and input from user. [Try max() function]

**Assingment 5.2:** WAP to add all numbers divisible by 7 and 9 from 1 to n and n is given by the user.

**Assingment 5.3:** WAP to add all prime numbers from 1 to n and n is given by the user.


````python
# Assignment 5.1
# WAP to find maximum among three numbers.
# Numbers are given by the user.

a = int(input("Enter First Number: "))
b = int(input("Enter Second Number: "))
c = int(input("Enter Third Number: "))

print("Maximum number =", max(a, b, c))


# Assignment 5.2
# WAP to add all numbers divisible by 7 and 9
# from 1 to n.
# Here, "divisible by 7 and 9" means divisible by both.

n = int(input("Enter n: "))

s = 0

for i in range(1, n + 1):
    if i % 7 == 0 and i % 9 == 0:
        s = s + i

print("Sum of numbers divisible by both 7 and 9 =", s)


# Assignment 5.3
# WAP to add all prime numbers from 1 to n.
# n is given by the user.

n = int(input("Enter n: "))

s = 0

for num in range(2, n + 1):
    is_prime = True

    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            is_prime = False
            break

    if is_prime:
        s = s + num

print("Sum of prime numbers from 1 to", n, "=", s)
````


**Assingment 6.1:** WAP using function that add all odd numbers from 1 to n, n is given by the user.

**Assingment 6.2:** WAP using function that add all prime numbers from 1 to n, n given by the user.

````python
# Assignment 6.1
# WAP using a function to add all odd numbers from 1 to n.
# n is given by the user.

def AddOdd(n):
    s = 0

    for i in range(1, n + 1):
        if i % 2 != 0:
            s = s + i

    return s


n = int(input("Enter n: "))
print("Sum of odd numbers from 1 to", n, "=", AddOdd(n))


# Assignment 6.2
# WAP using a function to add all prime numbers from 1 to n.
# n is given by the user.

def IsPrime(n):
    if n < 2:
        return False

    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False

    return True


def AddPrime(n):
    s = 0

    for i in range(2, n + 1):
        if IsPrime(i):
            s = s + i

    return s


n = int(input("Enter n: "))
print("Sum of prime numbers from 1 to", n, "=", AddPrime(n))
````

### **9.5 Generate random string**
---

````python
import string as s
import random as r
print ("String        --> ",s.ascii_letters)

passwd=r.sample(s.ascii_letters, 6)
print ("Selected Char --> ",passwd)

passwd1="".join(passwd)
print ("passwd1       --> ",passwd1)

passwd2="+".join(passwd)
print ("passwd2       --> ",passwd2)

passwd3="*".join(passwd)
print ("passwd3       --> ",passwd3)
````

---
### **9.6 Generate random digits**
---

````python
import string as s
import random as r
print ("Digits --> ",s.digits)

otp=r.sample(s.digits, 5)
print ("Selected num1 --> ",otp)
otp="".join(otp)
print ("otp1          --> ",otp)

otp=r.sample(s.digits, 5)
print ("Selected num2 --> ",otp)
otp="".join(otp)
print ("otp2          --> ",otp)

otp=r.sample(s.digits, 5)
print ("Selected num2 --> ",otp)
otp="".join(otp)
print ("otp3          --> ",otp)
````

### **9.7 Generate random string + digits**
---

````python
import string as s
import random as r
print ("String + Digits --> ",s.ascii_letters + s.digits)

mixPasswd=r.sample(s.ascii_letters + s.digits, 5)
print ("\nSelected Str1 --> ",mixPasswd)
mixPasswd="".join(mixPasswd)
print ("mixPasswd1    --> ",mixPasswd)

mixPasswd=r.sample(s.ascii_letters + s.digits, 6)
print ("\nSelected Str2 --> ",mixPasswd)
mixPasswd="".join(mixPasswd)
print ("mixPasswd2    --> ",mixPasswd)

splChar="#@!~%^&*()_+=-[]{}|"
mixPasswd=r.sample(splChar + s.ascii_letters + s.digits, 8)
print ("\nSelected Str3 --> ",mixPasswd)
mixPasswd="".join(mixPasswd)
print ("mixPasswd3    --> ",mixPasswd)
````
