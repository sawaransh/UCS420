Find the most frequent value and their indice(s) in the following arrays

```python
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])

values, counts = np.unique(x, return_counts=True)
most_frequent = values[np.argmax(counts)]
indices = np.where(x == most_frequent)[0]

print("Most frequent value:", most_frequent)
print("Indices:", indices)
```

Q3. For the given 2-D array arr=np.array([10, 20, 30], [40, 50, 60], [70, 80, 90]), access
elements using row and column indices as follows

```python
import numpy as np

arr = np.array([[10, 20, 30],
                [40, 50, 60],
                [70, 80, 90]])

print("Array:")
print(arr)

# a) Access 1st row, 2nd column
print("1st row, 2nd column:", arr[0, 1])

# b) Access 3rd row, 1st column
print("3rd row, 1st column:", arr[2, 0])
```

Q.4 Write program to create an 1-D NumPy array named <<Your Name>> with evenly
spaced 25 numbers from 10 to 100 using linspace(). Print the dimensions of the array,
shape, total elements, the data type of each element and total number of bytes consumed
by the array. Find the transpose of this array using reshape() attribute. Can we do the same
with T attribute?

```python
import numpy as np

# Create a 1-D array with 25 evenly spaced numbers from 10 to 100
Sawaransh = np.linspace(10, 100, 25)

print("Array:", Sawaransh)

# Dimensions of the array
print("Dimensions:", Sawaransh.ndim)

# Shape of the array
print("Shape:", Sawaransh.shape)

# Total number of elements
print("Total elements:", Sawaransh.size)

# Data type of each element
print("Data type:", Sawaransh.dtype)

# Total number of bytes consumed
print("Total bytes:", Sawaransh.nbytes)

# Transpose using reshape()
transpose_array = Sawaransh.reshape(1, 25).T

print("Transpose using reshape():")
print(transpose_array)

# Transpose using T attribute
print("Transpose using T attribute:")
print(Sawaransh.T)
```

Q5. Create a 2-D Array of three rows and four columns, named ucs420_<your_name>>
with following values – 10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35. Compute the mean,
median, max, min, unique elements. Reshape the array to four rows and three columns and
name it as reshaped_ ucs420_<your_name>>. Resize the array to two rows and three
columns and name it as resized_ ucs420_<your_name>>

```python
import numpy as np

# Create a 2-D array with 3 rows and 4 columns
ucs420_Sawaransh = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Original Array:")
print(ucs420_Sawaransh)

# Basic statistical operations
print("Mean:", np.mean(ucs420_Sawaransh))
print("Median:", np.median(ucs420_Sawaransh))
print("Maximum:", np.max(ucs420_Sawaransh))
print("Minimum:", np.min(ucs420_Sawaransh))
print("Unique elements:", np.unique(ucs420_Sawaransh))

# Reshape to 4 rows and 3 columns
reshaped_ucs420_Sawaransh = ucs420_Sawaransh.reshape(4, 3)

print("\nReshaped Array (4 x 3):")
print(reshaped_ucs420_Sawaransh)

# Resize to 2 rows and 3 columns
resized_ucs420_Sawaransh = np.resize(ucs420_Sawaransh, (2, 3))

print("\nResized Array (2 x 3):")
print(resized_ucs420_Sawaransh)
```
